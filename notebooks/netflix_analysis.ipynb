```python
rating_group = {
    'G': 'Kids',
    'TV-Y': 'Kids',
    'TV-Y7': 'Kids',
    'TV-G': 'Kids',
    'TV-Y7-FV': 'Kids',
    'PG': 'Teens',
    'TV-PG': 'Teens',
    'PG-13': 'Teens',
    'TV-14': 'Teens',
    'R': 'Adults',
    'TV-MA': 'Adults',
    'NC-17': 'Adults'
}

df['audience_group'] = df['rating'].map(rating_group)
```

```python
audience_counts = df['audience_group'].value_counts()

audience_counts.plot(kind='bar', figsize=(8, 5))

plt.title('Netflix Content by Audience Group')
plt.xlabel('Audience Group')
plt.ylabel('Number of Titles')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
audience_counts = df["audience_group"].value_counts()

audience_percentage = audience_counts/audience_counts.sum()*100

print(audience_percentage)
print(audience_percentage.round(2))
```

Audience Distribution - Buisness Insight

> Adults account for approximatley 45.99% of the Netflix Catalog.
> Teens content represents approximately 43.59% of the Catalog.
> Kids content represents only 10.42% of the total Catallog.
> Approximately 89.58% of the Catalog falls under the adults and teens.

Business Interpretation: Netflix's catalog is heavily concentrated in Adults and Teens categories, while Kids content represents a comparatively smaller share. This highlights the distribution of content across our simplified audience segments, but does not establish viewer demand or content performance.

```python
country_data = df['country'].dropna().str.split(',').explode().str.strip()

country_counts = country_data.value_counts()

print(country_counts.head(10))
```

```python
top_countries = country_counts.head(10)

top_countries.plot(kind='bar', figsize=(12, 6))

plt.title('Top 10 Countries by Netflix Content')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Country-wise Content Distribution – Business Insights

- The Netflix content catalog is highly concentrated in one leading country, which contributes significantly more titles than the other countries in the Top 10.
- The difference between the leading country and the remaining countries indicates an uneven distribution of content across geographical markets.
- Countries with comparatively fewer titles may represent potential opportunities for localized content expansion.
- Netflix can explore regional productions, local partnerships, and language-specific content to strengthen its catalog in different markets.

Business Interpretation: Netflix has a geographically concentrated content library. Expanding localized content may help diversify its catalog and better serve regional audiences.

Limitation: The number of titles available in a country does not directly measure viewership, popularity, customer demand, or revenue.

```python
movie_duration = df[df["type"] == "Movie"]["duration"]

movie_duration = movie_duration.str.replace(" min", "", regex=False)

movie_duration = movie_duration.astype(float)

movie_duration.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.hist(movie_duration, bins=20, edgecolor="black")

plt.title("Netflix Movie Duration Distribution")
plt.xlabel("Movie Duration (Minutes)")
plt.ylabel("Number of Movies")

plt.tight_layout()
plt.show()
```

```python
print("Average Movie Duration:", movie_duration.mean())
print("Median Movie Duration:", movie_duration.median())
print("Shortest Movie:", movie_duration.min())
print("Longest Movie:", movie_duration.max())
```

Movie Duration – Business Insight

- Most movies in the Netflix catalog are concentrated around the 90–120 minute range.
- Extremely short and very long movies are comparatively less common.
- This indicates that the catalog mainly contains standard-length movies.

Business Interpretation: Netflix's movie catalog is largely focused on standard-length films, which may suit viewers looking for conventional movie-watching experiences. However, catalog duration alone cannot determine viewer preference or engagement.
